# AulaPredict · Entrenamiento sintético reproducible

Abre el notebook con un kernel del entorno Python 3.11.15 del proyecto. Antes de ejecutarlo, instala `requirements.txt` desde la raíz del clon.

El pipeline genera sus propios rasgos faciales y objetivos académicos con semilla 42. Los resultados solo evalúan los datos sintéticos del generador.

```python
from pathlib import Path
import json
import subprocess
import sys

inicio = Path.cwd().resolve()
raiz = next((p for p in (inicio, *inicio.parents)
             if (p / 'backend/entrenamiento_prediccion/entrenar_prediccion.py').is_file()), None)
if raiz is None:
    raise RuntimeError('Abre el notebook desde el repositorio AulaPredict.')
salida = raiz / '_archivo' / 'entrenamiento_demo'
print(sys.version.split()[0], salida)
```

```python
subprocess.run([sys.executable, '-B', '-m',
                'backend.entrenamiento_prediccion.entrenar_prediccion',
                '--salida', str(salida)], cwd=raiz, check=True);
```

```python
metricas = json.loads((salida / 'metricas_entrenamiento.json').read_text(encoding='utf-8'))
print('Accuracy facial sintética:', metricas['modelo_facial_7_emociones']['accuracy'])
print('MAE de objetivos sintéticos:', metricas['modelo_sucesos_multimodal']['mae_global'])
print('R² de objetivos sintéticos:', metricas['modelo_sucesos_multimodal']['r2_score'])
print('Artefactos:', sorted(p.name for p in salida.iterdir()))
```

Los resultados se guardan en `_archivo/entrenamiento_demo/`, directorio ignorado por Git. El notebook no sustituye los pesos versionados.
